Problema 1: Desenpenho de uma rede de supermercados

```python
faturamentoSemana = []
faturamentoSemana.append (float(input("Digite o faturamento da segunda-feira: ")))
faturamentoSemana.append (float(input("Digite o faturamento da terça-feira: ")))
faturamentoSemana.append (float(input("Digite o faturamento da quarta-feira: ")))
faturamentoSemana.append (float(input("Digite o faturamento da quinta-feira: ")))
faturamentoSemana.append (float(input("Digite o faturamento da sexta-feira: ")))
faturamentoMedioDiario = (sum(faturamentoSemana))/(len(faturamentoSemana))
classFinalFaturamento = None

if faturamentoMedioDiario >= 22000:
    classFinalFaturamento = "Bom Desempenho"
elif faturamentoMedioDiario>=19000:
    classFinalFaturamento = "Desempenho Regular"
else:
    classFinalFaturamento = "Desempenho Crítico"

print(f"Resumo da semana de faturamento:\nQuantidade de dias analisados: {len(faturamentoSemana)}\nFaturamento total: {sum(faturamentoSemana)}\nFaturamento médio diário: {sum(faturamentoSemana)/len(faturamentoSemana)}\nMaior faturamento: {max(faturamentoSemana)}\nMenor faturamento: {min(faturamentoSemana)}\nClassificação final: {classFinalFaturamento}")
```

Problema 2: Controle de Velocidade

```python
velocidadeUser = int(input("Qual sua velocidade na passagem do radar? "))
velocidadeMax = 70
multa = 0
excessoPercentual = 0

if velocidadeUser>velocidadeMax:
    multa = (velocidadeUser-velocidadeMax)*9
    excessoPercentual = ((velocidadeUser-velocidadeMax)/velocidadeMax)*100

print(f"Sua velocidade foi de: {velocidadeUser}")
if velocidadeUser>velocidadeMax:
    print(f"Você passou da velocidade limite!\nVocê execedeu {round(excessoPercentual,2)}% da velocidade máxima, e custará R$ {multa} como multa.")
else:
    print("Você não excedeu o limite, Parabéns!")
```
