```python
notasAluno = [8.5,4.0,7.5,5.0,9.0]
nota6acima = []

for notaAluno in notasAluno:
    if notaAluno>=6:
        nota6acima.append(notaAluno)

print(f"As notas superiores a 6 foram:{nota6acima}")
print(f"A média destas notas é de:{round(sum(nota6acima)/len(nota6acima),2)}")
```

```python
tempos = [32,48,27,55,41,29]
tempMax = 40
countInd = 0
tempoTotal = 0
entregasAtrasadas = 0
entregasNAtrasadas = 0


for tempo in tempos:
    tempoTotal+=tempo
    countInd+=1
    if tempo>tempMax:
        entregasAtrasadas+=1
        print(f"A entrega {countInd} teve o tempo de entrega de {tempo} e estava fora do tempo máximo esperado da entrega.")
    else:
        entregasNAtrasadas+=1
        print(f"A entrega {countInd} teve o tempo de entrega de {tempo} e estava dentro do tempo máximo esperado da entrega.")

print(f"O tempo total das entregas foi de: {tempoTotal}\n O tempo médio das entregas foi de: {round(tempoTotal/len(tempos),2)}\n A quantidade de entregas atrasadas foi de: {entregasAtrasadas}\n E as dentro do prazo foi de: {entregasNAtrasadas}\n Totalizando {round((entregasAtrasadas/len(tempos))*100,2)}% de entregas atrasadas no total.")
```

```python
producoes = []
prodTotal = 0
qntdMeta = 0
qntdAbaixo = 0
qntdCrit = 0

inputUser = int(input("Digite o valor de peças produzidas entre 300 e 800 peças e digite -1 caso queira cancelar: "))
while inputUser != -1:
    producoes.append(inputUser)
    inputUser = int(input("Digite o valor de peças produzidas entre 300 e 800 peças e digite -1 caso queira cancelar: "))

for qntd in producoes:
    if qntd >= 500:
        qntdMeta+=1
    elif qntd >= 400:
        qntdAbaixo+=1
    else:
        qntdCrit+=1
    prodTotal+=qntd

print(f"O quantidade total de peças produzidas foi de: {prodTotal}\nA quantidade de turnos foi de: {len(producoes)}\nA média de peças produzidas foi de: {prodTotal/len(producoes):.2f}\nO total de turnos com produção dentro da meta foi de: {qntdMeta}\nTotalizando {(qntdMeta/len(producoes))*100:.2f}% de turnos abaixo da meta no total.")
    
```
