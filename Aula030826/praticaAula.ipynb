Problema Aplicado 1: Calculadora Básica

```python
print("Bem-vindo(a) à Calculadora Básica.\n Para utilizar, insira 2 valores numéricos e depois veja os resultados das operações.")
valorUm = float(input("Digite o primeiro valor: "))
valorDois = float(input("Digite o segundo valor: "))
print("A soma dos números é de:",valorUm + valorDois)
print("A diferença dos números é de:",valorUm - valorDois)
print("O produto dos números é de:",valorUm * valorDois)
print("A potencia dos números é de:",valorUm ** valorDois)
print("O quociente dos números é de:",valorUm / valorDois)
print("A divisão inteira dos números é de:",valorUm // valorDois)
print("O resto da divisão dos números é de:",valorUm % valorDois)
print("A raíz quadrada do primeiro número:",valorUm ** 0.5)
print("A raíz quadrada do segundo número:",valorDois ** 0.5)

```
