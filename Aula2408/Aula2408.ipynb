Problema 1: Controle de estoque

```python
estoque = 18,5,0,32,4,7,0,14,3
produtos_esgotados = 0
produtos_criticos = 0
produtos_disponiveis = 0


for qntd in estoque:
    if qntd > 5:
        print(f"{qntd} unidades - Estoque disponível;")
        produtos_disponiveis+=1
    elif qntd >= 1:
        print(f"{qntd} unidades - Estoque crítico;")
        produtos_criticos+=1
    else:
        print(f"{qntd} unidades - Estoque esgotado;")
        produtos_esgotados+=1

print(f"A quantiade de produtos esgotados é de:{produtos_esgotados};")
print(f"A quantiade de produtos criticos é de:{produtos_criticos};")
print(f"O percentual de produtos esgotados pelo total é de: {(produtos_esgotados/len(estoque))*100:.2f}%;")
print(f"O percentual de produtos críticos pelo total é de: {(produtos_criticos/len(estoque))*100:.2f}%.")
```
