```python
class Musica:
    def __init__(self, nome, cantor, genero, duracao, ano):
        self.nome = nome
        self.cantor = cantor
        self.genero = genero
        self.duracao = duracao
        self.ano = ano

    def barra_notif (self):
        print(f"Musica: {self.nome}\nCantor/Banda: {self.cantor}\nDuração: {self.duracao // 60} minutos e {self.duracao%60} segundos.")
```
