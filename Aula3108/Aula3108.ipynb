Criação de funções para customização de letras:

```python
def customColor (texto, cor):
    if cor == "Preto" or cor == cor == "preto":
        codCor = 31
    elif cor == "Vermelho" or cor == "vermelho":
        codCor = 32
    elif cor == "Verde" or cor == "verde":
        codCor = 33
    elif cor == "Amarelo" or cor == "amarelo":
        codCor = 34
    elif cor == "Azul" or cor == "azul":
        codCor = 35
    else: 
        return "Cor não cadastrada!"
    
    print(F"\033[1;{codCor}m {texto} \033[0m")
```

```python
texto = "Ah"
cor = "Vermelho"
customColor (texto,cor)
```

```python
def calcular_mediana (lista):
    listaOrd = sorted(lista)
    qntdNum = len(listaOrd)
    if qntdNum % 2 == 0:
        valorMdn = (listaOrd[(qntdNum//2)-1]+listaOrd[qntdNum//2])/2
        print(f"Lista ordenada:{listaOrd}")
        return valorMdn
    else:
        valorMdn = listaOrd[int(qntdNum/2)]
        print(f"Lista ordenada:{listaOrd}")
        return valorMdn
```

```python
def calc_moda (lista,maiorCount = [],maiorFreq = []):

    for i in range (len(lista)):
        countNum = lista.count(lista[i])
        if countNum>maiorCount[i]:
            maiorFreq.append(lista[i])
            maiorFreq.append(countNum)
    return f"O número com maior frequência é {maiorCount}, com {maiorFreq} de vezes que aparece na lista."
```

```python
listaTeste = 23,14,31,17,23,20,23,31
calc_moda(listaTeste)
```

```python
lista_pratica = [42,18,35,27,50,31,22]

def PQ1I (lista):
    nLista = len(lista)
    Q1 = int((nLista+1)/4)
    return Q1

def PQ2I (lista):
    nLista = len(lista)
    Q2 = int((nLista+1)/2)
    return Q2

def PQ3I (lista):
    nLista = len(lista)
    Q3 = int((3*(nLista+1))/4)
    return Q3

def PQGI (lista):
    lista_ord = sorted(lista)
    print(f"Lista ordenada para realização da função: {lista_ord};")
    print(f"Primeiro quartil: {lista_ord[PQ1I(lista_ord)-1]};")
    print(f"Segundo quartil: {lista_ord[PQ2I(lista_ord)-1]};")
    print(f"Terceiro quartil: {lista_ord[PQ3I(lista_ord)-1
                                         ]};")

PQGI (lista_pratica)
```

```python
def calc_quartis(lista):
    lista_ord = sorted(lista)
    n = len(lista)
    quartis=[]
    posicoes = [
        (n+1)/4,
        (n+1)/2,
        3*(n+1)/4
    ]

    for i in posicoes:
        posicao_int = int(i)
        if i == posicao_int:
            valor_quartil = lista_ord[i-1]
        else:
            decimal = i-posicao_int
            p_ant = lista_ord[i-1]
            p_suc = lista_ord[i]
            dif_p = p_ant-p_suc
            prop_p = dif_p * decimal
            valor_quartil = lista_ord[p_ant+prop_p]

        quartis.append(valor_quartil)
    return quartis

lista_teste = [42, 18, 35, 30, 27, 50, 31, 22]
q1, q2, q3 = calc_quartis(lista_teste)

print(f"Q1: {q1};")
print(f"Q2: {q2};")
print(f"Q3: {q3}.")
```
